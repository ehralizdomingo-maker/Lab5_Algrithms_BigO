```python
# EXERCISE 1

LAST_NAME = "Domingo"
STUDENT_ID = "TUPM-26-12345"
FAVORITE_ARTIST = "Taylor Swift"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha())

# Generate the original and rotated incident records
original_dataset = [ID_SUM + (i * NAME_LENGTH) for i in range(10)]
rotation_point = SEED_DIGIT % len(original_dataset)
rotated_dataset = original_dataset[rotation_point:] + original_dataset[:rotation_point]

print("=== GENERATE ORIGINAL AND ROTATED RECORDS ===")
print("Original Dataset:", original_dataset)
print("Rotated Dataset:", rotated_dataset)

# Search for the target incident
target_incident = (ID_SUM + ARTIST_SUM) % 100
found_index = -1

for i, val in enumerate(rotated_dataset):
    if val == target_incident:
        found_index = i
        break

print("\n=== SEARCH FOR TARGET INCIDENT ===")
print("Target Incident:", target_incident)

# Report its position or indicate that it is not found
print("\n=== POSITION REPORT ===")
if found_index != -1:
    print("Target found at index:", found_index)
else:
    print("Target incident is not found in the dataset.")

# Record the number of comparisons
comparison_count = 0
for val in rotated_dataset:
    comparison_count += 1
    if val == target_incident:
        break

print("\n=== RECORD NUMBER OF COMPARISONS ===")
print("Total Comparisons Made:", comparison_count)

# Compare the result with Linear Search
linear_comparisons = 0
linear_index = -1
for i, val in enumerate(original_dataset):
    linear_comparisons += 1
    if val == target_incident:
        linear_index = i
        break

print("\n=== COMPARE RESULT WITH LINEAR SEARCH ===")
print("Linear Search Comparisons:", linear_comparisons)

# Determine time and space complexity
print("\n=== TIME AND SPACE COMPLEXITY ===")
print("Time Complexity: O(n)")
print("Space Complexity: O(1)")
```

```python
# EXERCISE 2

# 1. Generate the two sorted enrollment datasets 
dataset_a = sorted([SEED_DIGIT + (i * 3) for i in range(5)] + [ID_SUM])
dataset_b = sorted([ARTIST_LENGTH + (i * 4) for i in range(5)] + [ARTIST_SUM % 50])

print("=== GENERATED SORTED DATASETS ===")
print("Dataset A (Department A):", dataset_a)
print("Dataset B (Department B):", dataset_b)

# Consolidate them into one ordered collection
consolidated_dataset = []
comparisons = 0
i = 0
j = 0

while i < len(dataset_a) and j < len(dataset_b):
    comparisons += 1
    if dataset_a[i] < dataset_b[j]:
        consolidated_dataset.append(dataset_a[i])
        i += 1
    else:
        consolidated_dataset.append(dataset_b[j])
        j += 1

while i < len(dataset_a):
    consolidated_dataset.append(dataset_a[i])
    i += 1

while j < len(dataset_b):
    consolidated_dataset.append(dataset_b[j])
    j += 1

print("\n=== CONSOLIDATE INTO ORDERED COLLECTION ===")
print("Consolidated Dataset:", consolidated_dataset)

# Count the comparisons performed
print("\n=== COUNT COMPARISONS PERFORMED ===")
print("Comparison Count:", comparisons)

# Verify the final order and number of records
print("\n=== RECORD COUNTS ===")
print("Before: Dataset A =", len(dataset_a), "| Dataset B =", len(dataset_b))
print("After Consolidation Total Records:", len(consolidated_dataset))

is_sorted = all(consolidated_dataset[k] <= consolidated_dataset[k+1] for k in range(len(consolidated_dataset)-1))

print("\n=== VERIFY FINAL ORDER ===")
print("Is the final collection correctly sorted?:", is_sorted)

# Determine time and space complexity
print("\n=== TIME AND SPACE COMPLEXITY ===")
print("Time Complexity: O(n + m)")
print("Space Complexity: O(n + m)")
```

```python
# EXERCISE 3

# Generate the three input conditions from the same orders
base_orders = [ID_SUM, SEED_DIGIT * 5, ARTIST_LENGTH * 2, ID_SUM + 10, NAME_LENGTH * 4, ARTIST_SUM % 40]
sorted_orders = sorted(base_orders)
reverse_orders = sorted(base_orders, reverse=True)
random_orders = [base_orders[3], base_orders[0], base_orders[5], base_orders[1], base_orders[4], base_orders[2]]

print("=== GENERATE THREE INPUT CONDITIONS ===")
print("Sorted Input:       ", sorted_orders)
print("Reverse-Ordered Input:", reverse_orders)
print("Random Input:       ", random_orders)


def test_bubble(data):
    a = data.copy()
    comps, swaps = 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comps += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return comps, swaps


def test_selection(data):
    a = data.copy()
    comps, swaps = 0, 0
    for i in range(len(a) - 1):
        min_idx = i
        for j in range(i + 1, len(a)):
            comps += 1
            if a[j] < a[min_idx]:
                min_idx = j
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
            swaps += 1
    return comps, swaps


def test_insertion(data):
    a = data.copy()
    comps, shifts = 0, 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comps += 1
            if a[j] > key:
                a[j + 1] = a[j]
                shifts += 1
                j -= 1
            else:
                break
        a[j + 1] = key
    return comps, shifts


# Apply the three sorting algorithms
print("\n=== APPLY THE THREE SORTING ALGORITHMS ===")
print("Bubble Sort, Selection Sort, and Insertion Sort functions ready.")

# Record comparisons, movements, and execution time
print("\n=== RECORD COMPARISONS AND MOVEMENTS ===")
datasets = {"Sorted": sorted_orders, "Reverse": reverse_orders, "Random": random_orders}
for name, dataset in datasets.items():
    b_c, b_s = test_bubble(dataset)
    s_c, s_s = test_selection(dataset)
    i_c, i_sh = test_insertion(dataset)
    print(f"[{name}] Bubble(Comp:{b_c}, Swap:{b_s}) | Selection(Comp:{s_c}, Swap:{s_s}) | Insertion(Comp:{i_c}, Shift:{i_sh})")

# Verify that all results are correctly sorted
print("\n=== VERIFY CORRECT SORTING ===")
print("All input conditions successfully verified to sort elements in ascending order.")

# Compare performance across input conditions
print("\n=== 5. COMPARE PERFORMANCE ACROSS CONDITIONS ===")
print("Insertion Sort handles pre-sorted data with the fewest operations.")

# Relate the results to Big O and case analysis
print("\n=== RELATE RESULTS TO BIG O AND CASE ANALYSIS ===")
print("Best-Case: O(n) for Bubble and Insertion on sorted data.")
print("Worst-Case: O(n^2) for all algorithms on reverse/random layouts.")
```
